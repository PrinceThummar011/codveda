**Clustering Analysis (K-
Means)**

```python
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df_cluster = df[['Retweets', 'Likes', 'Hour', 'Day']].dropna()
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df_cluster)
```

```python
inertia = []
k_range = range(1, 11)

for k in k_range:
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(X_scaled)
    inertia.append(kmeans.inertia_)

plt.figure(figsize=(8, 4))
plt.plot(k_range, inertia, marker='o')
plt.title('Elbow Method For Optimal K')
plt.xlabel('Number of Clusters')
plt.ylabel('Inertia')
plt.grid(True)
plt.show()
```

```python
# Let’s assume optimal k is 3
kmeans = KMeans(n_clusters=3, random_state=42)
df_cluster['Cluster'] = kmeans.fit_predict(X_scaled)
```

```python
plt.figure(figsize=(8, 6))
sns.scatterplot(data=df_cluster, x='Retweets', y='Likes', hue='Cluster', palette='Set1')
plt.title('K-Means Clustering (Retweets vs Likes)')
plt.xlabel('Retweets')
plt.ylabel('Likes')
plt.grid(True)
plt.show()
```

```python
df['Sentiment'] = df['Sentiment'].str.strip().str.lower()
```

```python
positive_words = [
    'positive', 'joy', 'happiness', 'love', 'amusement', 'enjoyment',
    'admiration', 'affection', 'awe', 'surprise', 'acceptance',
    'adoration', 'anticipation', 'hope', 'kind', 'pride', 'elation',
    'euphoria', 'contentment', 'serenity', 'gratitude', 'inspiration',
    'enthusiasm', 'wonder', 'confidence', 'success', 'creativity',
    'celebration', 'freedom', 'motivation', 'satisfaction', 'blessed',
    'cheerful', 'joyful', 'overjoyed', 'ecstasy', 'romance', 'playful',
    'whimsy', 'fun', 'delight'
]

negative_words = [
    'negative', 'anger', 'fear', 'sadness', 'disgust', 'disappointed',
    'shame', 'despair', 'grief', 'loneliness', 'frustration', 'envy',
    'regret', 'helplessness', 'anxiety', 'resentment', 'sorrow',
    'hate', 'bad', 'boredom', 'betrayal', 'heartbreak', 'loss',
    'darkness', 'desperation', 'pain', 'isolation', 'obstacle',
    'pressure', 'embarrassed'
]

neutral_words = [
    'neutral', 'confusion', 'calmness', 'numbness', 'indifference',
    'reflection', 'contemplation', 'solitude', 'energy', 'emotion'
]
```

```python
def map_sentiment(label):
    label = label.strip().lower()
    if label in positive_words:
        return 'Positive'
    elif label in negative_words:
        return 'Negative'
    elif label in neutral_words:
        return 'Neutral'
    else:
        return 'Neutral'
```

```python
df['Sentiment_Cleaned'] = df['Sentiment'].apply(map_sentiment)
print(df['Sentiment_Cleaned'].value_counts())
```
